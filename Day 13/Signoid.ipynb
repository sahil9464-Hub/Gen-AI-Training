```python
'''
With the help of linear
regression we can use function called sigmoid function

Sigmoid dunction will convert linear line into sigmoid value

Sigmoid value always 0 or 1

Sigmoid Value never exact 0 or 1

0.1 to 0.5 = 0
0.5 > 1 

Formula:--

Sigmoid(x) = 1 / (1 + e^(-x))   
(value of e - 2.718)
'''
# X_test , Y_pred :- Simple Model Score
# X_test , Y_test :- X_test value internally test/train then Score
# Y_test , Y_pred :- Compare behave Score generates (Classification ---- Accuracy)
```

```python
#Binary Classification using Linear and Logistic Regression
import numpy as np
x = np.random.random(30)
x
```

```python
plt.plot(x)
plt.show()
```

```python
# Create Sigmoid Function
def sigmoid_fun(x):
    return 1 / (1 + 2.71 ** (-x))
```

```python
sigmoid_x = sigmoid_fun(x)
plt.plot(sigmoid_x)
plt.show()
```

```python
new_x = np.linspace(-10, 10, 100)
sigmoid_x_new = sigmoid_fun(new_x)
plt.plot(sigmoid_x_new)
plt.xlabel("X")
plt.ylabel("Sigmoid(X)")
plt.title("Sigmoid Function")
plt.grid()
plt.show()
```

```python
#Building Model
model = LinearRegression()
model
```
