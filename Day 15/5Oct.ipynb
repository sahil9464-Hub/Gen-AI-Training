```python
# Last technique for data proprocessing using mean and Median
# 8th interview-
# why mean and median used for data preprocessing
# they work with numerical data only
# mean and median always working in data preprocessing for NaN value not for 0
import numpy as np 
import pandas as pd
```

```python
data={
    "age":[20,21,np.nan,22,21],
    "marks":[80,90,np.nan,75,np.nan]
    }
df=pd.DataFrame(data)
df
```

```python
df["age"]= df["age"].mean()
df
```

```python
df["marks"]= df["marks"].fillna(df["marks"].median())
df
```
