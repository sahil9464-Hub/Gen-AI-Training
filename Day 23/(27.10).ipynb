```python
from sklearn.preprocessing import OneHotEncoder
```

```python
df=pd.DataFrame({
    "City":["Delhi","Mumbai","Noida","Delhi","Mumbai","Gurgaon"]
})
df
```

```python
encoder=OneHotEncoder()
encoder
```

```python
encoder_array=encoder.fit_transform(df[["City"]])
encoder_array
```

```python
cols_name=encoder.get_feature_names_out(["City"])
cols_name
```

```python
result = pd.DataFrame(encoder_array.toarray(), columns=cols_name, dtype=int)
result
```

```python
cols_name=OneHotEncoder(sparse_output=False,handle_unknown="ignore")
cols_name
```

```python
df=pd.read_csv("loan_approval_dataset_200.csv")
df
```

```python
encoder=pd.get_dummies(df,dtype=int)
encoder
```

```python
categorical_cols = [
    'PropertyAreaCode',
    'Dependents'
]
```

```python
encoder = OneHotEncoder(
    sparse_output=False,
    handle_unknown='ignore'
)
```

```python
encoded_array = encoder.fit_transform(
    df[categorical_cols]
)
encoded_array
```

```python
encoded_df = pd.DataFrame(encoded_array)
```

```python
df = df.drop(columns=categorical_cols)
```

```python
df_final = pd.concat([df, encoded_df], axis=1)
```
