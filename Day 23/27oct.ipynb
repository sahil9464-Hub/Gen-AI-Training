```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
df=pd.DataFrame({
    "City":["Delhi","Mumbai","Delhi","Mumbai","Gurgaon"]
})
df
```

```python
encoder = OneHotEncoder()
encoder
```

```python
encoder_array=encoder.fit_transform(df[["City"]]).toarray()
encoder_array
```

```python
# get columns name from OneHotEncoder
cols_names=encoder.get_feature_names_out(["City"])
cols_names
```

```python
result=pd.DataFrame(encoder_array,columns=cols_names,dtype=int)
result
```

```python
df=pd.DataFrame({
    "Names":["Ravi","Suresh","Sita","Sonu","Gita","Mukesh","Rita"],
    "City":["Delhi","Mumbai","Delhi","Mumbai","Gurgaon","Delhi","Mumbai"],
    "Department":["HR","IT","HR","IT","HR","IT","HR"],
    "Age":[23,24,25,26,27,28,29]

})
df
```

```python
final = pd.concat([result,df[["Age"]].reset_index(drop=True)],axis=1)
final
```
