```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('house_prce.csv')
```

```python
# 2. Check for missing values
df.isnull().sum()
```

```python
# 3. Separate X and y
X = df[['Area_sqft', 'Bedrooms', 'Age_years']]
y = df['Price']
```

```python
# 4. Train Test Split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state = 3)
```

```python
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
# 5. Fill missing values using median
# find the median of each column from trainArea
Area_sqft_median = X_train['Area_sqft'].median()
Bedrooms_median = X_train['Bedrooms'].median()
Age_years_median = X_train['Age_years'].median()
```

```python
# fill missing values in train data
X_train['Area_sqft'] = X_train['Area_sqft'].fillna(Area_sqft_median)
X_train['Bedrooms'] = X_train['Bedrooms'].fillna(Bedrooms_median)
X_train['Age_years'] = X_train['Age_years'].fillna(Age_years_median)
```

```python
# fill missing values in test data (use the same train medians)
X_test['Area_sqft'] = X_test['Area_sqft'].fillna(Area_sqft_median)
X_test['Bedrooms'] = X_test['Bedrooms'].fillna(Bedrooms_median)
X_test['Age_years'] = X_test['Age_years'].fillna(Age_years_median)
```

```python
# 6. Check missing values again

print(X_train.isnull().sum())
print(X_test.isnull().sum())
```

```python
# 7. Create model

model = LinearRegression()
model
```

```python
# 8. Train model

model.fit(X_train, y_train)
```

```python
# 9. Prediction
y_pred = model.predict(X_test)

new_y_pred = np.round(y_pred, 2)
result = pd.DataFrame({'Actual Price': y_test.values, 'Predicted Price': new_y_pred})
result.head(10)
```

```python
# 10. Print coefficient and intercepts
print("Coefficients:", dict(zip(X.columns, model.coef_)))
print("Intercept:", model.intercept_)
```

```python
# 11. Actual vs Predicted Chart

plt.figure(figsize=(12, 6))

plt.plot(y_test.values[:50], color='red', marker='o', label='Actual Price')
plt.plot(y_pred[:50], color='green', marker='x', label='Predicted Price')

plt.xlabel('House Number')
plt.ylabel('Price (Crores)')
plt.title('Actual vs Predicted House Price')
plt.legend()
plt.grid(True)
plt.show()
```
